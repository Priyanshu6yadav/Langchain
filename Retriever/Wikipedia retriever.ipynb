## Wikipedia Retriever


```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
from langchain_community import retrievers
# Initialize the retriever (optional: set language and top_k)
retriever = WikipediaRetriever(top_k_results=2, lang='en') # hi for hindi language
```

```python
# query define
query = "the geopolitical history of india and pakistan from the perspective of the chinese"

# Get relevent  wikipedia doucuments
docs = retriever.invoke(query)
```

```python
print(docs)
```

```python
for i, doc in enumerate(docs):
  print(f"\n--- Resutl {i+1} ---")
  print(f"Content:\n{doc.page_content}...")
```
