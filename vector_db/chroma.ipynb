```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
```

```python
embeddings = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
```

```python
# create langchain documents of ipl players
doc1 = Document(
    page_content='Virat Kohli: The all-time leading run-scorer in IPL history, he holds the record for the most runs in a single season (973 runs in 2016) and has played his entire IPL career for Royal Challengers Bengaluru.',
    metadata = {"team":"Royal Challengers bangalore"}
)
doc2 = Document(
    page_content='Rohit Sharma: One of the most successful captains in IPL history, "The Hitman" led the Mumbai Indians to 5 IPL titles and has scored over 7,000 runs in the tournament.',
    metadata = {'team':'mumbai indians'}
)
doc3 = Document(
    page_content='MS Dhoni: Legendary captain of Chennai Super Kings who led his franchise to 5 IPL titles and holds the record for the most matches played as a captain and most dismissals as a wicketkeeper',
    metadata = {'team':'chennai super kings'}
)
doc4 = Document(
    page_content='Sachin Tendulkar: The first Indian player to win the IPL Orange Cap (in 2010), he represented Mumbai Indians from 2008 to 2013, winning the title in his final season',
    metadata={'team':'mumbai indians'}
)
doc5 = Document(
    page_content='Shubman Gill: A prominent modern-era opener who won the Orange Cap in 2023 with 890 runs and currently serves as the captain of Gujarat Titans',
    metadata = {'team':'gujarat titans'}
)
```

```python
docs = [doc1,doc2,doc3,doc4,doc5]

# creating database
vector_store = Chroma(
    embedding_function = embeddings,
    persist_directory = 'my_chroma_db',
    collection_name = 'sample'
)
```

```python
# add documents
vector_store.add_documents(docs)
```

```python
# view documents
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
# Search documents
vector_store.similarity_search(
    query='who among these are best captains',
    k=2 # to seen how much top result
)
```

```python
# search with similarity score
vector_store.similarity_search_with_score(
    query='who among these are best captains',
    k=2 # low score means more similarity
)
```

```python
# meta-data filtering
vector_store.similarity_search_with_score(
    query='',
    filter={'team':'mumbai indians'}
)
```

```python
# Keep one ID and delete the duplicate IDs
vector_store.delete(ids=[
    'c44478e3-9380-436a-9cc0-68c35ef9a78f',  # duplicate of Sachin Tendulkar
    '79a339c3-5f03-4330-a58c-472217b0ab72'   # duplicate of Rohit Sharma
])
```
